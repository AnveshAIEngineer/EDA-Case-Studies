#### 5. Create a histogram to show the distribution of the total_bill column. Add a title and appropriate labels.

```python
sns.histplot(tips['total_bill'])
plt.xlabel('Total Bill')
plt.title('Total Bill Distribution using Histogram')
plt.show()
```

#### 6. Use a bar plot to show the average tip amount for each day (day column). Add appropriate labels and a title

```python
import warnings
warnings.filterwarnings('ignore')
tip_mean = tips.groupby('day')['tip'].mean()
sns.barplot(x=tip_mean.index,y=tip_mean.values)
plt.xlabel('Days')
plt.ylabel('Tips')
plt.title('Average Tip Amount by Day')
plt.show()
```

#### 7. Create a box plot to show the distribution of tips (tip column) for different times of day (time column).

```python
sns.boxplot(x=tips['time'],y=tips['tip'])
plt.xlabel('Time')
plt.ylabel('Tips')
plt.title('Time and tips')
plt.show()
```

#### 8. Plot a violin plot to visualize the distribution of tips for each group size (size column).

```python
sns.violinplot(data=tips,y='tip',x='size')
```
